```r
# =========================
# 0) Load packages
# =========================
library(haven)
library(dplyr)
library(readxl)
library(ggplot2)
library(survival)
library(survminer)
library(scales)
library(tidyr)
library(writexl)
library(broom)
library(ggpubr)
library(mediation)


fmt_p <- function(p) {
  if (is.na(p)) return(NA_character_)
  if (p < 0.001) return("<0.001")
  sprintf("%.3f", p)
}

fmt_num <- function(x, digits = 1) {
  ifelse(
    is.na(x),
    NA_character_,
    format(round(x, digits), nsmall = digits, trim = TRUE)
  )
}

summarize_continuous <- function(df, var, by) {
  gvals <- df[[by]]
  x <- df[[var]]

  groups <- sort(unique(gvals[!is.na(gvals)]))

  out_stats <- lapply(seq_along(groups), function(i) {
    g <- groups[i]
    xv <- x[gvals == g]
    xv <- xv[!is.na(xv)]
    if (length(xv) == 0) return(NA_character_)
    paste0(
      fmt_num(median(xv, na.rm = TRUE), 1),
      " (",
      fmt_num(min(xv, na.rm = TRUE), 1),
      "–",
      fmt_num(max(xv, na.rm = TRUE), 1),
      ")"
    )
  })
  names(out_stats) <- paste0("stat_", seq_along(groups))

  tmp <- df[, c(var, by), drop = FALSE]
  tmp <- tmp[complete.cases(tmp), , drop = FALSE]

  pval <- NA_real_
  if (nrow(tmp) > 1 && length(unique(tmp[[by]])) >= 2) {
    if (length(unique(tmp[[by]])) == 2) {
      pval <- tryCatch(
        wilcox.test(tmp[[var]] ~ as.factor(tmp[[by]]))$p.value,
        error = function(e) NA_real_
      )
    } else {
      pval <- tryCatch(
        kruskal.test(tmp[[var]] ~ as.factor(tmp[[by]]))$p.value,
        error = function(e) NA_real_
      )
    }
  }

  tibble(
    variable = var,
    level = NA_character_,
    type = "continuous",
    N = sum(!is.na(x)),
    !!!out_stats,
    p = fmt_p(pval)
  )
}

summarize_categorical <- function(df, var, by) {
  tmp <- df[, c(var, by), drop = FALSE]
  tmp <- tmp[complete.cases(tmp), , drop = FALSE]

  if (nrow(tmp) == 0) {
    return(tibble(
      variable = var,
      level = NA_character_,
      type = "categorical",
      N = 0,
      stat_1 = NA_character_,
      p = NA_character_
    ))
  }

  tmp[[var]] <- as.character(tmp[[var]])
  tmp[[by]]  <- as.character(tmp[[by]])

  levels_var <- sort(unique(tmp[[var]]))
  groups <- sort(unique(tmp[[by]]))

  tab <- table(tmp[[var]], tmp[[by]], useNA = "no")

  out <- lapply(seq_along(levels_var), function(i) {
    lev <- levels_var[i]
    row <- list(
      variable = if (i == 1) var else "",
      level = lev,
      type = "categorical",
      N = sum(tmp[[var]] == lev, na.rm = TRUE)
    )

    for (j in seq_along(groups)) {
      g <- groups[j]
      n <- if (lev %in% rownames(tab) && g %in% colnames(tab)) tab[lev, g] else 0
      denom <- sum(tab[, g])
      row[[paste0("stat_", j)]] <- if (denom > 0) {
        paste0(n, " (", sprintf("%.1f", 100 * n / denom), "%)")
      } else {
        NA_character_
      }
    }
    row
  }) %>% bind_rows()

  pval <- NA_real_
  if (nrow(tab) > 0 && ncol(tab) > 1) {
    pval <- tryCatch({
      chi <- suppressWarnings(chisq.test(tab))
      if (all(chi$expected >= 5)) {
        chi$p.value
      } else {
        fisher.test(tab)$p.value
      }
    }, error = function(e) NA_real_)
  }

  out$p <- c(fmt_p(pval), rep("", nrow(out) - 1))
  out
}

build_summary_table <- function(df, vars, by, force_continuous = NULL) {
  res <- lapply(vars, function(v) {
    x <- df[[v]]
    treat_as_cont <- FALSE

    if (!is.null(force_continuous) && v %in% force_continuous) {
      treat_as_cont <- TRUE
    } else if (is.numeric(x)) {
      treat_as_cont <- TRUE
    }

    if (treat_as_cont) {
      summarize_continuous(df, v, by)
    } else {
      summarize_categorical(df, v, by)
    }
  })

  bind_rows(res)
}

rename_stat_columns <- function(tbl, labels) {
  stat_cols <- grep("^stat_", names(tbl), value = TRUE)
  if (length(stat_cols) != length(labels)) {
    stop("Number of labels must match number of stat columns.")
  }
  names(tbl)[match(stat_cols, names(tbl))] <- labels
  tbl
}

build_regression_table <- function(model, exponentiate = FALSE) {
  broom::tidy(model, conf.int = TRUE, exponentiate = exponentiate) %>%
    mutate(
      estimate = round(estimate, 3),
      conf.low = round(conf.low, 3),
      conf.high = round(conf.high, 3),
      p.value = vapply(p.value, fmt_p, character(1)),
      effect = paste0(estimate, " (", conf.low, ", ", conf.high, ")")
    ) %>%
    dplyr::select(term, effect, p.value)
}

# =========================================================
# Read data
# =========================================================
uni_df <- read_excel("./uni_df.xlsx")
nrow(uni_df) # 831

# =========================================================
# Baseline dFLC
# =========================================================
ggplot(uni_df, aes(x = dFLC)) + 
  geom_histogram(bins = 50) + 
  theme_classic()

df_sub <- uni_df %>%
  filter(!is.na(productive_IgH), !is.na(dFLC))

wilcox.test(dFLC ~ productive_IgH, data = df_sub)

ggplot(df_sub, aes(x = factor(productive_IgH), y = dFLC, fill = factor(productive_IgH))) +
  geom_violin(trim = FALSE, alpha = 0.6) +
  geom_boxplot(width = 0.1, outlier.shape = NA) +
  scale_y_log10() +
  stat_compare_means(method = "wilcox.test") +
  theme_classic()


# =========================================================
# Baseline table by productive_IgH
# =========================================================
desired_order <- c(
  "Sex", "Age", "Age≥65", "AL isotype (1=λ)", "productive_IgH", "t(11;14)","Sx_Dx",
  "Involved Organs (n)", "Organs>2", "Cardiac involvement", "Renal involvement", "Hepatic involvement",
  "BMPC (%)", "BMPC>10", "dFLC", "dFLC≥180",
  "Mayo04", "Mayo04III", "Mayo04IIIb","Mayo12","Mayo12III_IV",
  "NYHA", "NYHA3/4", "NT-ProBNP(pg/mL)", "NTproBNP>8500", "BNP (ng/L)",
  "cTnl (ug/L)", "cTnI>0.1", "LVPW (mm)", "IVS (mm)", "EF (%)", "EF < 50",
  "eGFR", "eGFR<50", "24hUP (g)", "24hUP≥5g", "Albumin (g/L)", "Alb<30",
  "treat", "Dara firstline", "Bort firstline", "Ven firstline"
)

baseline_vars <- intersect(desired_order, names(uni_df))
baseline_vars

tbl1 <- build_summary_table(
  df = uni_df,
  vars = baseline_vars,
  by = "productive_IgH",
  force_continuous = c("Involved Organs (n)")
)

tbl1

# =========================================================
# Baseline table by cardiac involvement
# =========================================================
desired_order <- c(
  "Sex", "Age", "Age≥65", "AL isotype (1=λ)", "productive_IgH", "t(11;14)",
  "Cardiac involvement", "Renal involvement", "Hepatic involvement",
  "BMPC (%)", "BMPC>10", "dFLC", "dFLC≥180",
  "eGFR", "eGFR<50", "24hUP (g)", "24hUP≥5g", "Albumin (g/L)", "Alb<30",
  "Dara firstline", "Bort firstline", "Ven firstline"
)

baseline_vars <- intersect(desired_order, names(uni_df))

tbl2 <- build_summary_table(
  df = uni_df,
  vars = baseline_vars,
  by = "Cardiac involvement"
)

tbl2

# =========================================================
# Create response variables
# =========================================================
uni_df <- uni_df %>%
  mutate(
    across(
      c(`3m HR`,`6m HR`,`12m HR`,`24m HR`),
      ~ ifelse(. %in% c("CR","VGPR","PR"), 1,
               ifelse(. %in% c("SD"), 0, NA)),
      .names = "{gsub(' HR','',.col)}_or"
    ),
    across(
      c(`3m HR`,`6m HR`,`12m HR`,`24m HR`),
      ~ ifelse(. %in% c("CR","VGPR"), 1,
               ifelse(. %in% c("PR","SD"), 0, NA)),
      .names = "{gsub(' HR','',.col)}_vgpr"
    ),
    across(
      c(`3m HR`,`6m HR`,`12m HR`,`24m HR`),
      ~ ifelse(. %in% c("CR"), 1,
               ifelse(. %in% c("VGPR","PR","SD"), 0, NA)),
      .names = "{gsub(' HR','',.col)}_cr_only"
    ),
    across(
      starts_with("cardiac_"),
      ~ ifelse(. %in% c("CR","VGPR","PR"), 1,
               ifelse(. %in% c("SD"), 0, NA)),
      .names = "car_{.col}_or"
    ),
    across(
      starts_with("cardiac_"),
      ~ ifelse(. %in% c("CR","VGPR"), 1,
               ifelse(. %in% c("PR","SD"), 0, NA)),
      .names = "car_{.col}_vgpr"
    )
  )

# =========================================================
# Response summary table by productive_IgH
# =========================================================
response_vars <- c(
  "best_hema_resp", "hema_or", "hema_vgpr", "hema_cr_only",
  "3m_or","6m_or","12m_or","24m_or",
  "3m_vgpr","6m_vgpr","12m_vgpr","24m_vgpr",
  "3m_cr_only","6m_cr_only","12m_cr_only","24m_cr_only",
  "car_cardiac_12m_or","car_cardiac_24m_or",
  "car_cardiac_12m_vgpr","car_cardiac_24m_vgpr"
)

response_vars <- intersect(response_vars, names(uni_df))

tbl3 <- build_summary_table(
  df = uni_df,
  vars = response_vars,
  by = "productive_IgH"
)

tbl3 <- tbl3 %>%
  mutate(
    variable = recode(
      variable,
      "best_hema_resp" = "Best Heme Response",
      "hema_cr_only" = "CR Rate",
      "hema_or" = "ORR",
      "hema_vgpr" = "≥VGPR rate"
    )
  )

# rename group columns if there are exactly 2 groups
stat_cols_tbl3 <- grep("^stat_", names(tbl3), value = TRUE)
if (length(stat_cols_tbl3) == 2) {
  tbl3 <- rename_stat_columns(
    tbl3,
    labels = c("LC only (N=407)", "IgH prod (N=373)")
  )
}

tbl3

# =========================================================
# Hematologic response stacked bar plot
# =========================================================
plot_long <- uni_df %>%
  filter(!is.na(`productive_IgH`)) %>%   
  dplyr::select(`productive_IgH`, `3m HR`, `6m HR`, `12m HR`, `24m HR`) %>%
  tidyr::pivot_longer(
    cols = c(`3m HR`, `6m HR`, `12m HR`, `24m HR`),
    names_to = "Time",
    values_to = "Response",
    names_pattern = "(.*) HR"
  ) %>%
  dplyr::mutate(
    Time = factor(Time, levels = c("3m", "6m", "12m", "24m")),
    Response = factor(Response, levels = c("PD","SD","PR","VGPR","CR"))
  )

plot_summary <- plot_long %>%
  filter(!is.na(Response)) %>%
  group_by(`productive_IgH`, Time, Response) %>%
  summarise(n = n(), .groups = "drop") %>%
  group_by(`productive_IgH`, Time) %>%
  mutate(
    total = sum(n),
    percent = n / total * 100
  ) %>%
  ungroup()

plot_n_orr <- plot_long %>%
  filter(!is.na(Response)) %>%
  group_by(`productive_IgH`, Time) %>%
  summarise(
    n = n(),
    ORR = mean(Response %in% c("CR","VGPR","PR")) * 100,
    .groups = "drop"
  )

ggplot(plot_summary, aes(x = Time, y = percent, fill = Response)) +
  geom_bar(stat = "identity", color = "black") +
  geom_text(
    aes(label = sprintf("%.0f%%", percent)),
    position = position_stack(vjust = 0.5),
    color = "white",
    size = 3.5,
    fontface = "bold"
  ) +
  geom_text(
    data = plot_n_orr,
    aes(x = Time, y = 104, label = sprintf("n = %d\nORR = %.1f%%", n, ORR)),
    inherit.aes = FALSE,
    size = 4,
    fontface = "bold",
    lineheight = 0.9
  ) +
  scale_fill_manual(
    values = c(
      "CR"   = "#1a9850",
      "VGPR" = "#91cf60",
      "PR"   = "#fee08b",
      "SD"   = "#fc8d59",
      "PD"   = "#d73027"
    ),
    name = "Hema Response"
  ) +
  facet_wrap(
    ~`productive_IgH`,
    labeller = labeller(
      `productive_IgH` = c(
        "0" = "LC",
        "1" = "IgH"
      )
    )
  ) +
  coord_cartesian(ylim = c(0, 106)) +
  theme_classic()



# =========================================================
# Regression and association analysis
# =========================================================
uni_df$log_dFLC <- log(uni_df$dFLC)

df2 <- uni_df %>% 
  dplyr::select(`Cardiac involvement`, productive_IgH, log_dFLC) %>% 
  na.omit()

lm(log(`NT-ProBNP(pg/mL)`) ~ productive_IgH + log_dFLC, data = uni_df)

summary(glm(`Cardiac involvement` ~ productive_IgH, data = df2, family = binomial))
summary(glm(`Cardiac involvement` ~ log_dFLC, data = df2, family = binomial))
summary(glm(`Cardiac involvement` ~ productive_IgH + log_dFLC, data = df2, family = binomial))
summary(lm(log_dFLC ~ productive_IgH, data = df2))

m1 <- glm(
  `Cardiac involvement` ~ 
    Age +
    productive_IgH +
    `BMPC>10` +
    log_dFLC +
    `eGFR<50` +
    `Alb<30` +
    `Dara firstline`,
  data = uni_df,
  family = binomial
)

table_m1 <- build_regression_table(m1, exponentiate = TRUE)
table_m1

# =========================================================
# Subset check
# =========================================================
df3 <- subset(uni_df, productive_IgH == 1)
nrow(df3)

summary(glm(`Cardiac involvement` ~ log_dFLC, data = df2, family = binomial))

```

```r
library(dplyr)
library(tidyr)
library(purrr)
library(broom)
library(stringr)

# -----------------------------
# helper: pretty p-value
# -----------------------------
fmt_p <- function(p) {
  if (is.na(p)) return(NA_character_)
  if (p < 0.001) return("<0.001")
  sprintf("%.3f", p)
}

# -----------------------------
# helper: safe tests
# -----------------------------
get_p_cont <- function(data, var, by = "productive_IgH") {
  df <- data %>%
    dplyr::select(all_of(c(by, var))) %>%
    filter(!is.na(.data[[by]]), !is.na(.data[[var]]))

  if (nrow(df) == 0) return(NA_real_)
  if (length(unique(df[[by]])) < 2) return(NA_real_)
  if (length(unique(df[[var]])) < 2) return(NA_real_)

  out <- tryCatch({
    wilcox.test(df[[var]] ~ df[[by]])$p.value
  }, error = function(e) NA_real_)

  out
}

get_p_cat <- function(data, var, by = "productive_IgH") {
  df <- data %>%
    dplyr::select(all_of(c(by, var))) %>%
    filter(!is.na(.data[[by]]), !is.na(.data[[var]]))

  if (nrow(df) == 0) return(NA_real_)
  if (length(unique(df[[by]])) < 2) return(NA_real_)
  if (length(unique(df[[var]])) < 2) return(NA_real_)

  tab <- table(df[[var]], df[[by]])

  if (any(dim(tab) < 2)) return(NA_real_)

  out <- tryCatch({
    if (any(tab < 5)) {
      fisher.test(tab)$p.value
    } else {
      suppressWarnings(chisq.test(tab)$p.value)
    }
  }, error = function(e) NA_real_)

  out
}

# -----------------------------
# helper: detect variable type
# treat numeric with only 0/1 as categorical
# -----------------------------
detect_var_type <- function(x) {
  if (is.numeric(x)) {
    ux <- unique(x[!is.na(x)])
    if (length(ux) <= 2 && all(sort(ux) %in% c(0, 1))) {
      return("categorical")
    } else {
      return("continuous")
    }
  } else {
    return("categorical")
  }
}

# -----------------------------
# helper: summarize one continuous variable
# -----------------------------
summarise_continuous <- function(data, var, by = "productive_IgH", labels = NULL) {
  group_vals <- sort(unique(data[[by]][!is.na(data[[by]])]))

  stat_list <- lapply(group_vals, function(g) {
    x <- data %>%
      filter(.data[[by]] == g) %>%
      pull(all_of(var))

    tibble(
      group = as.character(g),
      value = if (all(is.na(x))) {
        NA_character_
      } else {
        sprintf(
          "%.1f (%.1f–%.1f)",
          median(x, na.rm = TRUE),
          min(x, na.rm = TRUE),
          max(x, na.rm = TRUE)
        )
      }
    )
  })

  stat_df <- bind_rows(stat_list) %>%
    pivot_wider(names_from = group, values_from = value)

  pval <- get_p_cont(data, var, by)

  tibble(
    Variable = if (!is.null(labels) && var %in% names(labels)) labels[[var]] else var,
    Level = ""
  ) %>%
    bind_cols(stat_df) %>%
    mutate(
      p_value = fmt_p(pval),
      sig = ifelse(!is.na(pval) & pval < 0.05, "*", "")
    )
}

# -----------------------------
# helper: summarize one categorical variable
# -----------------------------
summarise_categorical <- function(data, var, by = "productive_IgH", labels = NULL) {
  df <- data %>%
    filter(!is.na(.data[[var]]), !is.na(.data[[by]])) %>%
    mutate(
      .var = as.character(.data[[var]]),
      .by  = as.character(.data[[by]])
    )

  if (nrow(df) == 0) {
    return(
      tibble(
        Variable = if (!is.null(labels) && var %in% names(labels)) labels[[var]] else var,
        Level = NA_character_,
        p_value = NA_character_,
        sig = ""
      )
    )
  }

  group_vals <- sort(unique(df$.by))
  lvl_vals <- unique(df$.var)

  counts <- df %>%
    count(.var, .by, name = "n") %>%
    group_by(.by) %>%
    mutate(pct = 100 * n / sum(n)) %>%
    ungroup() %>%
    mutate(stat = sprintf("%d (%.1f%%)", n, pct)) %>%
    dplyr::select(.var, .by, stat) %>%
    pivot_wider(names_from = .by, values_from = stat)

  pval <- get_p_cat(data, var, by)

  out <- counts %>%
    rename(Level = .var) %>%
    mutate(
      Variable = if (!is.null(labels) && var %in% names(labels)) labels[[var]] else var,
      p_value = ""
    ) %>%
    relocate(Variable, Level)

  if (nrow(out) > 0) {
    out$p_value[1] <- fmt_p(pval)
  }

  out <- out %>%
    mutate(
      sig = ""
    )

  if (nrow(out) > 0 && !is.na(pval) && pval < 0.05) {
    out$sig[1] <- "*"
  }

  out
}

# -----------------------------
# main table builder
# -----------------------------
make_summary_table <- function(data, vars, by = "productive_IgH", labels = NULL) {
  vars <- intersect(vars, names(data))

  pieces <- lapply(vars, function(v) {
    vtype <- detect_var_type(data[[v]])
    if (vtype == "continuous") {
      summarise_continuous(data, v, by, labels)
    } else {
      summarise_categorical(data, v, by, labels)
    }
  })

  out <- bind_rows(pieces)

  # add group Ns
  n_df <- data %>%
    filter(!is.na(.data[[by]])) %>%
    count(.data[[by]], name = "N") %>%
    mutate(group = as.character(.data[[by]]))

  group_vals <- sort(unique(data[[by]][!is.na(data[[by]])]))

  # rename group columns more nicely
  old_names <- intersect(as.character(group_vals), names(out))
  new_names <- paste0(old_names, " (N=", n_df$N[match(old_names, n_df$group)], ")")
  names(out)[match(old_names, names(out))] <- new_names

  out
}
```

```r
library(ggplot2)
library(dplyr)
library(tidyr)
library(viridisLite)

magma_cols <- viridisLite::magma(11)

# =========================================================
# 1) Build patient-level long response data from uni_df
# =========================================================
plot_long <- uni_df %>%
  filter(!is.na(productive_IgH)) %>%
  dplyr::select(productive_IgH, `3m HR`, `6m HR`, `12m HR`, `24m HR`) %>%
  pivot_longer(
    cols = c(`3m HR`, `6m HR`, `12m HR`, `24m HR`),
    names_to = "Time",
    values_to = "Response",
    names_pattern = "(.*) HR"
  ) %>%
  mutate(
    Time = factor(Time, levels = c("3m", "6m", "12m", "24m")),
    productive_IgH = as.character(productive_IgH),
    productive_IgH = recode(productive_IgH, "0" = "LC", "1" = "IgH"),
    productive_IgH = factor(productive_IgH, levels = c("LC", "IgH")),
    Response = factor(Response, levels = c("PD", "SD", "PR", "VGPR", "CR"))
  )

# =========================================================
# 2) Patient-level CR binary for proper testing
#    CR = 1, VGPR/PR/SD/PD = 0, missing stays NA
# =========================================================
cr_patient <- plot_long %>%
  mutate(
    CR_binary = case_when(
      Response == "CR" ~ 1,
      Response %in% c("VGPR", "PR", "SD", "PD") ~ 0,
      TRUE ~ NA_real_
    )
  ) %>%
  filter(!is.na(CR_binary), !is.na(productive_IgH), !is.na(Time))

# =========================================================
# 3) Wilcoxon test by Time on patient-level binary CR
#    (mathematically valid as a rank test on 0/1 values,
#     though Fisher/chi-square is usually more natural for binary outcomes)
# =========================================================
pval_df <- cr_patient %>%
  group_by(Time) %>%
  summarise(
    n_LC  = sum(productive_IgH == "LC"),
    n_IgH = sum(productive_IgH == "IgH"),
    CR_LC_pct  = mean(CR_binary[productive_IgH == "LC"],  na.rm = TRUE) * 100,
    CR_IgH_pct = mean(CR_binary[productive_IgH == "IgH"], na.rm = TRUE) * 100,
    p_value = tryCatch(
      wilcox.test(CR_binary ~ productive_IgH)$p.value,
      error = function(e) NA_real_
    ),
    .groups = "drop"
  ) %>%
  mutate(
    label = ifelse(
      is.na(p_value),
      "Wilcoxon p = NA",
      paste0("Wilcoxon p = ", signif(p_value, 3))
    )
  )

print(pval_df)

# =========================================================
# 4) Aggregated percentages for stacked bar plot
# =========================================================
plot_summary <- plot_long %>%
  filter(!is.na(Response)) %>%
  group_by(productive_IgH, Time, Response) %>%
  summarise(n = n(), .groups = "drop") %>%
  group_by(productive_IgH, Time) %>%
  mutate(
    total = sum(n),
    percent = n / total * 100
  ) %>%
  ungroup()

# n per facet/time
plot_n <- plot_long %>%
  filter(!is.na(Response)) %>%
  group_by(productive_IgH, Time) %>%
  summarise(
    n = n(),
    .groups = "drop"
  )

# =========================================================
# 5) Annotation data: repeat p-values into both facets
# =========================================================
y_top  <- max(plot_summary$percent, na.rm = TRUE)
y_n    <- y_top + 7
y_p    <- y_top + 14
y_lim  <- y_top + 22

pval_annot <- bind_rows(
  pval_df %>% mutate(productive_IgH = factor("LC", levels = c("LC", "IgH"))),
  pval_df %>% mutate(productive_IgH = factor("IgH", levels = c("LC", "IgH")))
)

plot_n_annot <- plot_n %>%
  mutate(label = paste0("n = ", n))

# =========================================================
# 6) Plot
# =========================================================
p <- ggplot(plot_summary, aes(x = Time, y = percent, fill = Response)) +
  geom_bar(stat = "identity", color = "black", width = 0.78) +
  geom_text(
    aes(label = sprintf("%.0f%%", percent)),
    position = position_stack(vjust = 0.5),
    color = "black",
    size = 4.0,
    fontface = "bold",
    lineheight = 1.0
  ) +
  geom_text(
    data = plot_n_annot,
    aes(x = Time, y = y_n, label = label),
    inherit.aes = FALSE,
    size = 4.0,
    fontface = "bold",
    vjust = 0
  ) +
  geom_text(
    data = pval_annot,
    aes(x = Time, y = y_p, label = label),
    inherit.aes = FALSE,
    size = 4.0,
    fontface = "bold",
    vjust = 0
  ) +
  scale_fill_manual(
    values = c(
      "CR"   = magma_cols[11],
      "VGPR" = magma_cols[8],
      "PR"   = magma_cols[6],
      "SD"   = magma_cols[3],
      "PD"   = magma_cols[1]
    ),
    name = "Hema Response"
  ) +
  facet_wrap(~ productive_IgH, nrow = 1) +
  coord_cartesian(ylim = c(0, y_lim), clip = "off") +
  labs(x = NULL, y = "Percent") +
  theme_classic(base_size = 15) +
  theme(
    strip.text = element_text(face = "bold"),
    axis.text.x = element_text(angle = 0, hjust = 0.5),
    plot.margin = margin(10, 10, 22, 10)
  )

p

ggsave("df_stack_with_patientlevel_wilcox.pdf", plot = p, height = 6, width = 8)
```

```r
# ============================================================
# Mediation analysis + Nature-style visualization
# Works directly from med.out produced by mediation::mediate()
# ============================================================

library(dplyr)
library(ggplot2)
library(patchwork)
library(grid)
library(mediation)

# ------------------------------------------------------------
# 1. Run mediation analysis
# ------------------------------------------------------------
uni_df$log_dFLC <- log(uni_df$dFLC)

df2 <- uni_df %>%
  dplyr::select(`Cardiac involvement`, productive_IgH, log_dFLC) %>%
  na.omit()

med.model <- lm(
  log_dFLC ~ productive_IgH,
  data = df2
)

out.model <- glm(
  `Cardiac involvement` ~ productive_IgH + log_dFLC,
  data = df2,
  family = binomial
)

med.out <- mediate(
  model.m  = med.model,
  model.y  = out.model,
  treat    = "productive_IgH",
  mediator = "log_dFLC",
  boot     = TRUE,
  sims     = 1000
)

summary(med.out)

# ------------------------------------------------------------
# 2. Extract mediation results correctly from med.out
# ------------------------------------------------------------
# For a standard binary treatment mediation analysis, med.out stores:
# ACME = d.avg
# ADE  = z.avg
# Total effect = tau.coef
# Proportion mediated = n.avg
#
# Bootstrap CIs:
# ACME: d.avg.ci
# ADE:  z.avg.ci
# Total: tau.ci
# Prop mediated: n.avg.ci

med_df <- data.frame(
  effect = c("ACME", "ADE", "Total effect", "Prop. mediated"),
  estimate = c(
    unname(med.out$d.avg),
    unname(med.out$z.avg),
    unname(med.out$tau.coef),
    unname(med.out$n.avg)
  ),
  lower = c(
    unname(med.out$d.avg.ci[1]),
    unname(med.out$z.avg.ci[1]),
    unname(med.out$tau.ci[1]),
    unname(med.out$n.avg.ci[1])
  ),
  upper = c(
    unname(med.out$d.avg.ci[2]),
    unname(med.out$z.avg.ci[2]),
    unname(med.out$tau.ci[2]),
    unname(med.out$n.avg.ci[2])
  )
)

eff_df <- med_df %>%
  filter(effect %in% c("ACME", "ADE", "Total effect")) %>%
  mutate(effect = factor(effect, levels = c("Total effect", "ADE", "ACME")))

prop_df <- med_df %>%
  filter(effect == "Prop. mediated")

acme_val  <- med_df$estimate[med_df$effect == "ACME"]
ade_val   <- med_df$estimate[med_df$effect == "ADE"]
total_val <- med_df$estimate[med_df$effect == "Total effect"]
prop_val  <- med_df$estimate[med_df$effect == "Prop. mediated"]

acme_ci  <- med_df %>% filter(effect == "ACME") %>% dplyr::select(lower, upper)
ade_ci   <- med_df %>% filter(effect == "ADE") %>% dplyr::select(lower, upper)
total_ci <- med_df %>% filter(effect == "Total effect") %>% dplyr::select(lower, upper)
prop_ci  <- med_df %>% filter(effect == "Prop. mediated") %>% dplyr::select(lower, upper)

# Optional p-values, if present in med.out
get_safe <- function(x) {
  if (is.null(x)) return(NA_real_)
  unname(x)
}

p_acme <- get_safe(med.out$d.avg.p)
p_ade  <- get_safe(med.out$z.avg.p)
p_tot  <- get_safe(med.out$tau.p)
p_prop <- get_safe(med.out$n.avg.p)

fmt_num <- function(x, digits = 3) sprintf(paste0("%.", digits, "f"), x)
fmt_pct <- function(x, digits = 1) sprintf(paste0("%.", digits, "f%%"), x * 100)
fmt_p <- function(p) {
  if (is.na(p)) return("p = NA")
  if (p < 0.001) return("p < 0.001")
  paste0("p = ", sprintf("%.3f", p))
}

# ------------------------------------------------------------
# 3. Panel A: mediation diagram
# ------------------------------------------------------------
pA <- ggplot() +
  xlim(0, 10) +
  ylim(0, 6.6) +
  theme_void() +

  # nodes
  annotate(
    "label", x = 1.5, y = 3.0, label = "X\nproductive_IgH",
    size = 5.0, label.size = 0.4, fill = "white"
  ) +
  annotate(
    "label", x = 5.0, y = 4.9, label = "M\nlog(dFLC)",
    size = 5.0, label.size = 0.4, fill = "white"
  ) +
  annotate(
    "label", x = 8.5, y = 3.0, label = "Y\nCardiac involvement",
    size = 5.0, label.size = 0.4, fill = "white"
  ) +

  # arrows
  annotate(
    "segment", x = 2.3, xend = 4.1, y = 3.5, yend = 4.5,
    linewidth = 0.8,
    arrow = arrow(length = unit(0.18, "inches"), type = "closed")
  ) +
  annotate(
    "segment", x = 5.9, xend = 7.7, y = 4.5, yend = 3.5,
    linewidth = 0.8,
    arrow = arrow(length = unit(0.18, "inches"), type = "closed")
  ) +
  annotate(
    "segment", x = 2.3, xend = 7.7, y = 2.35, yend = 2.35,
    linewidth = 0.8,
    arrow = arrow(length = unit(0.18, "inches"), type = "closed")
  ) +

  # path labels
  annotate("text", x = 3.2, y = 4.62, label = "a path", size = 4.4) +
  annotate("text", x = 6.8, y = 4.62, label = "b path", size = 4.4) +
  annotate("text", x = 5.0, y = 2.72, label = "Direct effect (ADE)", size = 4.2) +

  # proportion mediated
  annotate(
    "text", x = 5.0, y = 5.95,
    label = paste0(fmt_pct(prop_val, 1), " mediated"),
    size = 4.3, fontface = "bold"
  ) +

  # result summary
  annotate(
    "text", x = 5.0, y = 1.02,
    label = paste0(
      "Indirect effect (ACME) = ", fmt_num(acme_val), 
      " [", fmt_num(acme_ci$lower), ", ", fmt_num(acme_ci$upper), "]\n",
      "Direct effect (ADE) = ", fmt_num(ade_val), 
      " [", fmt_num(ade_ci$lower), ", ", fmt_num(ade_ci$upper), "]\n",
      "Total effect = ", fmt_num(total_val), 
      " [", fmt_num(total_ci$lower), ", ", fmt_num(total_ci$upper), "]\n",
      "Proportion mediated = ", fmt_pct(prop_val, 1),
      " [", fmt_pct(prop_ci$lower, 1), ", ", fmt_pct(prop_ci$upper, 1), "]"
    ),
    size = 4.0, lineheight = 1.15
  ) +

  annotate(
    "text", x = 0.2, y = 6.45, label = "A",
    hjust = 0, vjust = 1, size = 7, fontface = "bold"
  )

# ------------------------------------------------------------
# 4. Panel B: forest plot for ACME/ADE/Total
# ------------------------------------------------------------
label_df <- eff_df %>%
  mutate(
    label_txt = sprintf("%.3f", estimate),
    label_x = min(lower, na.rm = TRUE) - 0.06 * diff(range(c(lower, upper), na.rm = TRUE))
  )

pB <- ggplot(eff_df, aes(x = estimate, y = effect)) +
  geom_vline(xintercept = 0, linetype = "dashed", linewidth = 0.5, color = "grey50") +
  geom_errorbarh(aes(xmin = lower, xmax = upper), height = 0.16, linewidth = 0.8) +
  geom_point(size = 3.2, shape = 21, fill = "black", color = "black", stroke = 0.6) +
  geom_text(
    data = label_df,
    aes(x = label_x, y = effect, label = label_txt),
    inherit.aes = FALSE,
    hjust = 1,
    size = 3.8
  ) +
  scale_x_continuous(expand = expansion(mult = c(0.20, 0.08))) +
  labs(x = "Effect size (95% bootstrap CI)", y = NULL) +
  theme_classic(base_size = 13) +
  theme(
    axis.text.y = element_text(size = 12, color = "black"),
    axis.text.x = element_text(size = 11, color = "black"),
    axis.title.x = element_text(size = 12),
    plot.margin = margin(10, 15, 10, 10)
  ) +
  annotate(
    "text", x = -Inf, y = Inf, label = "B",
    hjust = -0.7, vjust = 1.7, size = 7, fontface = "bold"
  )

# ------------------------------------------------------------
# 5. Panel C: proportion mediated with correct CI scale
# ------------------------------------------------------------
ymin_c <- min(0, prop_df$lower, prop_df$estimate, na.rm = TRUE)
ymax_c <- max(0, prop_df$upper, prop_df$estimate, na.rm = TRUE)
ypad_c <- 0.08 * (ymax_c - ymin_c)
if (ypad_c == 0) ypad_c <- 0.1

pC <- ggplot(prop_df, aes(x = effect, y = estimate)) +
  geom_col(width = 0.52, fill = "grey20") +
  geom_errorbar(aes(ymin = lower, ymax = upper), width = 0.12, linewidth = 0.8) +
  geom_hline(yintercept = 0, linewidth = 0.5) +
  coord_cartesian(ylim = c(ymin_c - ypad_c, ymax_c + ypad_c)) +
  labs(x = NULL, y = "Proportion mediated") +
  theme_classic(base_size = 13) +
  theme(
    axis.text.x = element_text(size = 11, color = "black"),
    axis.text.y = element_text(size = 11, color = "black"),
    axis.title.y = element_text(size = 12),
    plot.margin = margin(10, 10, 10, 10)
  ) +
  annotate(
    "text", x = 1, y = prop_df$estimate + 0.04 * (ymax_c - ymin_c + 1e-8),
    label = fmt_pct(prop_df$estimate, 1),
    size = 4
  ) +
  annotate(
    "text", x = -Inf, y = Inf, label = "C",
    hjust = -0.7, vjust = 1.7, size = 7, fontface = "bold"
  )

# ------------------------------------------------------------
# 6. Combine panels
# ------------------------------------------------------------
final_plot <- pA + pB
  plot_annotation(
    title = "Causal mediation analysis",
    theme = theme(
      plot.title = element_text(size = 15, face = "bold", hjust = 0.5)
    )
  )

final_plot

# ------------------------------------------------------------
# 7. Save
# ------------------------------------------------------------
ggsave(
  "mediation_nature_style.pdf",
  final_plot,
  width = 10,
  height = 5,
  useDingbats = FALSE
)

ggsave(
  "mediation_nature_style.png",
  final_plot,
  width = 10,
  height = 5,
  dpi = 400
)
```

```r
# =========================
# 0) Load packages
# =========================
suppressPackageStartupMessages({
  library(dplyr)
  library(survival)
  library(survminer)
  library(ggplot2)
})

# =========================
# 0.5) Output directory + helpers
# =========================
outdir <- "plots_pdf"
dir.create(outdir, showWarnings = FALSE, recursive = TRUE)

save_ggsurvplot_pdf <- function(x, filename, width = 7, height = 7) {
  pdf(file.path(outdir, filename), width = width, height = height, onefile = FALSE)
  print(x)
  dev.off()
}

save_ggplot_pdf <- function(x, filename, width = 7, height = 7) {
  ggsave(
    filename = file.path(outdir, filename),
    plot = x,
    device = cairo_pdf,
    width = width,
    height = height,
    units = "in"
  )
}

# =========================
# 1) Clean column names
# =========================
dat <- uni_df %>%
  rename(
    cardiac = `Cardiac involvement`,
    dflc = dFLC,
    cr_3m = `3m_cr_only`,
    sex = Sex,
    age = Age
  )

# =========================
# 2) Make variables usable
# =========================
dat <- dat %>%
  mutate(
    productive_IgH = as.character(productive_IgH),
    cr_3m = as.character(cr_3m),
    sex = factor(sex),
    cardiac = as.character(cardiac),
    dflc = suppressWarnings(as.numeric(dflc)),
    age = suppressWarnings(as.numeric(age)),
    os_time = suppressWarnings(as.numeric(os_time)),
    os_event = suppressWarnings(as.numeric(os_event))
  )

# Label productive_IgH
# 0 = LC-only
# 1 = LC-with-IgH
if (all(na.omit(unique(dat$productive_IgH)) %in% c("0", "1"))) {
  dat <- dat %>%
    mutate(
      productive_IgH = factor(
        productive_IgH,
        levels = c("1", "0"),
        labels = c("LC-with-IgH", "LC-only")
      )
    )
} else {
  dat <- dat %>% mutate(productive_IgH = factor(productive_IgH))
}

# Label CR
if (all(na.omit(unique(dat$cr_3m)) %in% c("0", "1"))) {
  dat <- dat %>%
    mutate(
      cr_3m = factor(
        cr_3m,
        levels = c("0", "1"),
        labels = c("No CR at 3m", "CR at 3m")
      )
    )
} else {
  dat <- dat %>% mutate(cr_3m = factor(cr_3m))
}

# Label cardiac involvement
if (all(na.omit(unique(dat$cardiac)) %in% c("0", "1"))) {
  dat <- dat %>%
    mutate(
      cardiac = factor(
        cardiac,
        levels = c("0", "1"),
        labels = c("No cardiac involvement", "Cardiac involvement")
      )
    )
} else {
  dat <- dat %>% mutate(cardiac = factor(cardiac))
}

# Optional: log-transform dFLC for Cox model
# If dFLC can be zero, use log1p; otherwise use log
dat <- dat %>%
  mutate(
    log_dflc = ifelse(!is.na(dflc), log(dflc), NA_real_)
  )

# =========================
# 3) Construct 3-month landmark dataset
# =========================
landmark_month <- 2.95

dat_lm <- dat %>%
  filter(
    !is.na(os_time),
    !is.na(os_event),
    !is.na(productive_IgH),
    !is.na(cr_3m),
    !is.na(cardiac),
    !is.na(log_dflc),
    !is.na(age),
    !is.na(sex)
  ) %>%
  # keep patients alive and under follow-up at 3 months
  filter(os_time >= landmark_month) %>%
  mutate(
    # redefine follow-up from the landmark
    os_time_landmark = os_time - landmark_month
  )

# Check sample size after landmarking
cat("N in landmark cohort:", nrow(dat_lm), "\n")
cat("Events in landmark cohort:", sum(dat_lm$os_event, na.rm = TRUE), "\n")

table(dat_lm$cr_3m, useNA = "ifany")
table(dat_lm$productive_IgH, useNA = "ifany")

# =========================
# 4) Set reference levels
# =========================
if ("LC-with-IgH" %in% levels(dat_lm$productive_IgH)) {
  dat_lm$productive_IgH <- relevel(dat_lm$productive_IgH, ref = "LC-with-IgH")
}

if ("No CR at 3m" %in% levels(dat_lm$cr_3m)) {
  dat_lm$cr_3m <- relevel(dat_lm$cr_3m, ref = "No CR at 3m")
}

if ("No cardiac involvement" %in% levels(dat_lm$cardiac)) {
  dat_lm$cardiac <- relevel(dat_lm$cardiac, ref = "No cardiac involvement")
}

# =========================
# 5) Kaplan-Meier from landmark
# =========================
fit_lm_cr <- survfit(
  Surv(os_time_landmark, os_event) ~ cr_3m,
  data = dat_lm
)

p_lm_cr <- ggsurvplot(
  fit_lm_cr, data = dat_lm,
  title = "Overall survival from 3-month landmark by 3-month CR",
  xlab = "Time since 3-month landmark (months)",
  ylab = "Probability",
  conf.int = FALSE,
  pval = TRUE,
  risk.table = TRUE,
  risk.table.height = 0.25,
  risk.table.y.text = FALSE,
  risk.table.fontsize = 4,
  legend = "right",
  break.time.by = 24,
  ggtheme = theme_classic(base_size = 14),
  palette = c("firebrick", "steelblue")
)

print(p_lm_cr)
save_ggsurvplot_pdf(p_lm_cr, "KM_OS_3m_landmark_by_CR.pdf", width = 7, height = 7)

# Optional: subgroup KMs from landmark
dat_lm_lc_only <- dat_lm %>% filter(productive_IgH == "LC-only")
dat_lm_lc_igh  <- dat_lm %>% filter(productive_IgH == "LC-with-IgH")

if (nrow(dat_lm_lc_only) > 0 && length(unique(dat_lm_lc_only$cr_3m)) > 1) {
  fit_lm_lc_only <- survfit(
    Surv(os_time_landmark, os_event) ~ cr_3m,
    data = dat_lm_lc_only
  )

  p_lm_lc_only <- ggsurvplot(
    fit_lm_lc_only, data = dat_lm_lc_only,
    title = "OS from 3-month landmark by CR in LC-only",
    xlab = "Time since 3-month landmark (months)",
    ylab = "Probability",
    conf.int = FALSE,
    pval = TRUE,
    risk.table = TRUE,
    risk.table.height = 0.25,
    risk.table.y.text = FALSE,
    risk.table.fontsize = 4,
    legend = "right",
    break.time.by = 24,
    ggtheme = theme_classic(base_size = 14),
    palette = c("firebrick", "steelblue")
  )

  print(p_lm_lc_only)
  save_ggsurvplot_pdf(p_lm_lc_only, "KM_OS_3m_landmark_by_CR_LC_only.pdf", width = 7, height = 7)
}

if (nrow(dat_lm_lc_igh) > 0 && length(unique(dat_lm_lc_igh$cr_3m)) > 1) {
  fit_lm_lc_igh <- survfit(
    Surv(os_time_landmark, os_event) ~ cr_3m,
    data = dat_lm_lc_igh
  )

  p_lm_lc_igh <- ggsurvplot(
    fit_lm_lc_igh, data = dat_lm_lc_igh,
    title = "OS from 3-month landmark by CR in LC-with-IgH",
    xlab = "Time since 3-month landmark (months)",
    ylab = "Probability",
    conf.int = FALSE,
    pval = TRUE,
    risk.table = TRUE,
    risk.table.height = 0.25,
    risk.table.y.text = FALSE,
    risk.table.fontsize = 4,
    legend = "right",
    break.time.by = 24,
    ggtheme = theme_classic(base_size = 14),
    palette = c("firebrick", "steelblue")
  )

  print(p_lm_lc_igh)
  save_ggsurvplot_pdf(p_lm_lc_igh, "KM_OS_3m_landmark_by_CR_LC_with_IgH.pdf", width = 7, height = 7)
}

# =========================
# 6) Univariable landmark Cox models
# =========================
cox_uni_igh_lm <- coxph(
  Surv(os_time_landmark, os_event) ~ productive_IgH,
  data = dat_lm
)

cox_uni_cr_lm <- coxph(
  Surv(os_time_landmark, os_event) ~ cr_3m,
  data = dat_lm
)

cox_uni_cardiac_lm <- coxph(
  Surv(os_time_landmark, os_event) ~ cardiac,
  data = dat_lm
)

cox_uni_dflc_lm <- coxph(
  Surv(os_time_landmark, os_event) ~ log_dflc,
  data = dat_lm
)

cox_uni_age_lm <- coxph(
  Surv(os_time_landmark, os_event) ~ age,
  data = dat_lm
)

cox_uni_sex_lm <- coxph(
  Surv(os_time_landmark, os_event) ~ sex,
  data = dat_lm
)

summary(cox_uni_igh_lm)
summary(cox_uni_cr_lm)
summary(cox_uni_cardiac_lm)
summary(cox_uni_dflc_lm)
summary(cox_uni_age_lm)
summary(cox_uni_sex_lm)

# =========================
# 7) Multivariable landmark Cox model
# =========================
cox_multi_lm <- coxph(
  Surv(os_time_landmark, os_event) ~ productive_IgH + cardiac + log_dflc + cr_3m + sex + age,
  data = dat_lm
)

summary(cox_multi_lm)

# =========================
# 8) Hazard ratio table
# =========================
cox_sum <- summary(cox_multi_lm)

hr_table <- data.frame(
  Variable = rownames(cox_sum$coefficients),
  HR = cox_sum$coefficients[, "exp(coef)"],
  Lower95 = cox_sum$conf.int[, "lower .95"],
  Upper95 = cox_sum$conf.int[, "upper .95"],
  Pvalue = cox_sum$coefficients[, "Pr(>|z|)"],
  row.names = NULL,
  check.names = FALSE
)

hr_table_fmt <- hr_table %>%
  mutate(
    HR_CI = sprintf("%.2f (%.2f-%.2f)", HR, Lower95, Upper95),
    Pvalue = ifelse(Pvalue < 0.001, "<0.001", sprintf("%.3f", Pvalue))
  ) %>%
  dplyr::select(Variable, HR_CI, Pvalue)

print(hr_table_fmt, row.names = FALSE)

# =========================
# 9) Forest plot
# =========================
plot_df <- hr_table %>%
  mutate(
    Label = sprintf("%.2f (%.2f-%.2f)", HR, Lower95, Upper95),
    Ptxt = ifelse(Pvalue < 0.001, "<0.001", sprintf("%.3f", Pvalue)),
    Variable = gsub("productive_IgH", "IgH subgroup: ", Variable),
    Variable = gsub("cardiac", "Cardiac: ", Variable),
    Variable = gsub("cr_3m", "3m CR: ", Variable),
    Variable = gsub("sex", "Sex: ", Variable),
    Variable = gsub("log_dflc", "log(dFLC)", Variable),
    Variable = gsub("age", "Age", Variable)
  ) %>%
  mutate(Variable = factor(Variable, levels = rev(Variable)))

xmax <- max(plot_df$Upper95, na.rm = TRUE)
xmin <- min(plot_df$Lower95, na.rm = TRUE)

p_forest <- ggplot(plot_df, aes(x = HR, y = Variable)) +
  geom_point(size = 3) +
  geom_errorbarh(aes(xmin = Lower95, xmax = Upper95), height = 0.2) +
  geom_vline(xintercept = 1, linetype = 2) +
  scale_x_log10(
    limits = c(min(0.5, xmin * 0.9), max(2, xmax * 2.2))
  ) +
  geom_text(aes(x = xmax * 1.35, label = Label), hjust = 0, size = 4) +
  geom_text(aes(x = xmax * 2.0, label = Ptxt), hjust = 0, size = 4) +
  annotate("text", x = xmax * 1.35, y = length(levels(plot_df$Variable)) + 0.8,
           label = "HR (95% CI)", hjust = 0, size = 4.2) +
  annotate("text", x = xmax * 2.0, y = length(levels(plot_df$Variable)) + 0.8,
           label = "P", hjust = 0, size = 4.2) +
  coord_cartesian(clip = "off") +
  labs(
    title = "3-month landmark multivariable Cox model for OS",
    x = "Hazard ratio (log scale)",
    y = NULL
  ) +
  theme_classic(base_size = 14) +
  theme(
    plot.margin = margin(5.5, 120, 5.5, 5.5)
  )

print(p_forest)
save_ggplot_pdf(p_forest, "Forest_3m_landmark_multivariable_Cox_OS.pdf", width = 7, height = 7)

# =========================
# 10) Proportional hazards assumption
# =========================
ph_test <- cox.zph(cox_multi_lm)
print(ph_test)

pdf(file.path(outdir, "PH_assumption_cox_zph_plots.pdf"), width = 7, height = 7, onefile = TRUE)
plot(ph_test)
dev.off()

# =========================
# 11) Optional interaction model
# =========================
cox_interaction_lm <- coxph(
  Surv(os_time_landmark, os_event) ~ productive_IgH * cr_3m + cardiac + log_dflc + sex + age,
  data = dat_lm
)

summary(cox_interaction_lm)
anova(cox_multi_lm, cox_interaction_lm, test = "LRT")
```

```r
suppressPackageStartupMessages({
  library(dplyr)
})

# -------------------------
# 1) Variables used in your landmark / Cox analysis
# -------------------------
vars_check <- c(
  "os_time",
  "os_event",
  "productive_IgH",
  "3m_cr_only",
  "Cardiac involvement",
  "dFLC",
  "Sex",
  "Age"
)

# keep only variables that actually exist
vars_check <- vars_check[vars_check %in% colnames(uni_df)]

cat("Variables checked:\n")
print(vars_check)

cat("\nOriginal cohort size:", nrow(uni_df), "\n")

# -------------------------
# 2) Missing count for each variable
# -------------------------
missing_summary <- data.frame(
  Variable = vars_check,
  N_missing = sapply(vars_check, function(v) sum(is.na(uni_df[[v]]))),
  Pct_missing = sapply(vars_check, function(v) mean(is.na(uni_df[[v]])) * 100),
  row.names = NULL
)

missing_summary <- missing_summary %>%
  arrange(desc(N_missing))

cat("\n=== Missingness by variable ===\n")
print(missing_summary)

# -------------------------
# 3) Sequential filtering audit
# -------------------------
filter_steps <- list(
  "Has os_time" = !is.na(uni_df$os_time),
  "Has os_event" = !is.na(uni_df$os_event),
  "Has productive_IgH" = !is.na(uni_df$productive_IgH),
  "Has 3m_cr_only" = !is.na(uni_df$`3m_cr_only`),
  "Has Cardiac involvement" = !is.na(uni_df$`Cardiac involvement`),
  "Has dFLC" = !is.na(uni_df$dFLC),
  "Has Sex" = !is.na(uni_df$Sex),
  "Has Age" = !is.na(uni_df$Age)
)

# keep only valid steps for columns that exist
filter_steps <- filter_steps[names(filter_steps) %in% paste("Has", vars_check)]

keep <- rep(TRUE, nrow(uni_df))

seq_table <- data.frame(
  Step = "Original cohort",
  N_remaining = nrow(uni_df),
  N_removed_at_step = 0
)

for (nm in names(filter_steps)) {
  before <- sum(keep)
  keep <- keep & filter_steps[[nm]]
  after <- sum(keep)

  seq_table <- rbind(
    seq_table,
    data.frame(
      Step = nm,
      N_remaining = after,
      N_removed_at_step = before - after
    )
  )
}

cat("\n=== Sequential filtering audit ===\n")
print(seq_table, row.names = FALSE)

# -------------------------
# 4) Who gets excluded from final landmark/Cox cohort
# -------------------------
required_all <- Reduce(`&`, filter_steps)

cat("\nFinal complete-case cohort size:", sum(required_all), "\n")
cat("Excluded from final cohort:", sum(!required_all), "\n")

# -------------------------
# 5) For each excluded patient, count how many fields are missing
# -------------------------
na_matrix <- sapply(vars_check, function(v) is.na(uni_df[[v]]))
if (is.vector(na_matrix)) na_matrix <- matrix(na_matrix, ncol = 1, dimnames = list(NULL, vars_check))

excluded_detail <- data.frame(
  row_id = seq_len(nrow(uni_df)),
  excluded = !required_all,
  n_missing_among_model_vars = rowSums(na_matrix)
)

cat("\n=== Missing fields among excluded patients ===\n")
print(table(excluded_detail$n_missing_among_model_vars[excluded_detail$excluded], useNA = "ifany"))

# -------------------------
# 6) Which variable is most often missing among excluded patients
# -------------------------
excluded_missing_by_var <- data.frame(
  Variable = vars_check,
  Missing_among_excluded = sapply(vars_check, function(v) {
    sum(is.na(uni_df[[v]]) & !required_all)
  }),
  Pct_of_excluded = sapply(vars_check, function(v) {
    mean(is.na(uni_df[[v]])[!required_all]) * 100
  }),
  row.names = NULL
) %>%
  arrange(desc(Missing_among_excluded))

cat("\n=== Missingness among excluded patients ===\n")
print(excluded_missing_by_var)

# -------------------------
# 7) Exact exclusion pattern combinations
# -------------------------
pattern_df <- as.data.frame(na_matrix)
colnames(pattern_df) <- vars_check

pattern_table <- pattern_df %>%
  mutate(
    exclusion_pattern = apply(
      .,
      1,
      function(x) {
        miss <- vars_check[as.logical(x)]
        if (length(miss) == 0) "Complete"
        else paste(miss, collapse = " | ")
      }
    )
  ) %>%
  count(exclusion_pattern, sort = TRUE)

cat("\n=== Most common exclusion patterns ===\n")
print(head(pattern_table, 20))

# -------------------------
# 8) Optional: see actual rows excluded
# -------------------------
excluded_rows <- uni_df[!required_all, , drop = FALSE]

cat("\nExcluded row count:", nrow(excluded_rows), "\n")

# Preview useful columns if they exist
preview_cols <- intersect(
  c("os_time", "os_event", "productive_IgH", "3m_cr_only",
    "Cardiac involvement", "dFLC", "Sex", "Age"),
  colnames(excluded_rows)
)

print(head(excluded_rows[, preview_cols, drop = FALSE], 20))
```
