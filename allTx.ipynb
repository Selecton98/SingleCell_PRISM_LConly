```r
# =========================================================
# Packages
# =========================================================
library(haven)
library(dplyr)
library(readxl)
library(ggplot2)
library(survival)
library(survminer)
library(scales)
library(tidyr)
library(writexl)
library(broom)
library(ggpubr)
library(mediation)

# =========================================================
# Helper functions replacing gtsummary
# =========================================================
fmt_p <- function(p) {
  if (is.na(p)) return(NA_character_)
  if (p < 0.001) return("<0.001")
  sprintf("%.3f", p)
}

fmt_num <- function(x, digits = 1) {
  ifelse(
    is.na(x),
    NA_character_,
    format(round(x, digits), nsmall = digits, trim = TRUE)
  )
}

summarize_continuous <- function(df, var, by) {
  gvals <- df[[by]]
  x <- df[[var]]

  groups <- sort(unique(gvals[!is.na(gvals)]))

  out_stats <- lapply(seq_along(groups), function(i) {
    g <- groups[i]
    xv <- x[gvals == g]
    xv <- xv[!is.na(xv)]
    if (length(xv) == 0) return(NA_character_)
    paste0(
      fmt_num(median(xv, na.rm = TRUE), 1),
      " (",
      fmt_num(min(xv, na.rm = TRUE), 1),
      "–",
      fmt_num(max(xv, na.rm = TRUE), 1),
      ")"
    )
  })
  names(out_stats) <- paste0("stat_", seq_along(groups))

  tmp <- df[, c(var, by), drop = FALSE]
  tmp <- tmp[complete.cases(tmp), , drop = FALSE]

  pval <- NA_real_
  if (nrow(tmp) > 1 && length(unique(tmp[[by]])) >= 2) {
    if (length(unique(tmp[[by]])) == 2) {
      pval <- tryCatch(
        wilcox.test(tmp[[var]] ~ as.factor(tmp[[by]]))$p.value,
        error = function(e) NA_real_
      )
    } else {
      pval <- tryCatch(
        kruskal.test(tmp[[var]] ~ as.factor(tmp[[by]]))$p.value,
        error = function(e) NA_real_
      )
    }
  }

  tibble(
    variable = var,
    level = NA_character_,
    type = "continuous",
    N = sum(!is.na(x)),
    !!!out_stats,
    p = fmt_p(pval)
  )
}

summarize_categorical <- function(df, var, by) {
  tmp <- df[, c(var, by), drop = FALSE]
  tmp <- tmp[complete.cases(tmp), , drop = FALSE]

  if (nrow(tmp) == 0) {
    return(tibble(
      variable = var,
      level = NA_character_,
      type = "categorical",
      N = 0,
      stat_1 = NA_character_,
      p = NA_character_
    ))
  }

  tmp[[var]] <- as.character(tmp[[var]])
  tmp[[by]]  <- as.character(tmp[[by]])

  levels_var <- sort(unique(tmp[[var]]))
  groups <- sort(unique(tmp[[by]]))

  tab <- table(tmp[[var]], tmp[[by]], useNA = "no")

  out <- lapply(seq_along(levels_var), function(i) {
    lev <- levels_var[i]
    row <- list(
      variable = if (i == 1) var else "",
      level = lev,
      type = "categorical",
      N = sum(tmp[[var]] == lev, na.rm = TRUE)
    )

    for (j in seq_along(groups)) {
      g <- groups[j]
      n <- if (lev %in% rownames(tab) && g %in% colnames(tab)) tab[lev, g] else 0
      denom <- sum(tab[, g])
      row[[paste0("stat_", j)]] <- if (denom > 0) {
        paste0(n, " (", sprintf("%.1f", 100 * n / denom), "%)")
      } else {
        NA_character_
      }
    }
    row
  }) %>% bind_rows()

  pval <- NA_real_
  if (nrow(tab) > 0 && ncol(tab) > 1) {
    pval <- tryCatch({
      chi <- suppressWarnings(chisq.test(tab))
      if (all(chi$expected >= 5)) {
        chi$p.value
      } else {
        fisher.test(tab)$p.value
      }
    }, error = function(e) NA_real_)
  }

  out$p <- c(fmt_p(pval), rep("", nrow(out) - 1))
  out
}

build_summary_table <- function(df, vars, by, force_continuous = NULL) {
  res <- lapply(vars, function(v) {
    x <- df[[v]]
    treat_as_cont <- FALSE

    if (!is.null(force_continuous) && v %in% force_continuous) {
      treat_as_cont <- TRUE
    } else if (is.numeric(x)) {
      treat_as_cont <- TRUE
    }

    if (treat_as_cont) {
      summarize_continuous(df, v, by)
    } else {
      summarize_categorical(df, v, by)
    }
  })

  bind_rows(res)
}

rename_stat_columns <- function(tbl, labels) {
  stat_cols <- grep("^stat_", names(tbl), value = TRUE)
  if (length(stat_cols) != length(labels)) {
    stop("Number of labels must match number of stat columns.")
  }
  names(tbl)[match(stat_cols, names(tbl))] <- labels
  tbl
}

build_regression_table <- function(model, exponentiate = FALSE) {
  broom::tidy(model, conf.int = TRUE, exponentiate = exponentiate) %>%
    mutate(
      estimate = round(estimate, 3),
      conf.low = round(conf.low, 3),
      conf.high = round(conf.high, 3),
      p.value = vapply(p.value, fmt_p, character(1)),
      effect = paste0(estimate, " (", conf.low, ", ", conf.high, ")")
    ) %>%
    dplyr::select(term, effect, p.value)
}

# =========================================================
# Read data
# =========================================================
uni_df <- read_excel("./uni_df.xlsx")
nrow(uni_df) # 831

# =========================================================
# Baseline dFLC
# =========================================================
ggplot(uni_df, aes(x = dFLC)) + 
  geom_histogram(bins = 50) + 
  theme_classic()

df_sub <- uni_df %>%
  filter(!is.na(productive_IgH), !is.na(dFLC))

wilcox.test(dFLC ~ productive_IgH, data = df_sub)

ggplot(df_sub, aes(x = factor(productive_IgH), y = dFLC, fill = factor(productive_IgH))) +
  geom_violin(trim = FALSE, alpha = 0.6) +
  geom_boxplot(width = 0.1, outlier.shape = NA) +
  scale_y_log10() +
  stat_compare_means(method = "wilcox.test") +
  theme_classic()

#medians <- df_sub %>%
#  group_by(M_type_bin) %>%
#  summarise(med = median(dFLC, na.rm = TRUE), .groups = "drop")

#ggplot(df_sub, aes(x = factor(M_type_bin), y = dFLC, fill = factor(M_type_bin))) +
#  geom_violin(trim = FALSE, alpha = 0.6) +
#  geom_boxplot(width = 0.1, outlier.shape = NA) +
#  stat_summary(fun = median, geom = "point", size = 3, color = "red") +
#  geom_text(
#    data = medians,
#    aes(x = factor(M_type_bin), y = med, label = round(med, 1)),
#    vjust = -1, size = 4
#  ) +
#  scale_y_log10() +
#  theme_classic()

# =========================================================
# Baseline table by productive_IgH
# =========================================================
desired_order <- c(
  "Sex", "Age", "Age≥65", "AL isotype (1=λ)", "productive_IgH", "t(11;14)","Sx_Dx",
  "Involved Organs (n)", "Organs>2", "Cardiac involvement", "Renal involvement", "Hepatic involvement",
  "BMPC (%)", "BMPC>10", "dFLC", "dFLC≥180",
  "Mayo04", "Mayo04III", "Mayo04IIIb","Mayo12","Mayo12III_IV",
  "NYHA", "NYHA3/4", "NT-ProBNP(pg/mL)", "NTproBNP>8500", "BNP (ng/L)",
  "cTnl (ug/L)", "cTnI>0.1", "LVPW (mm)", "IVS (mm)", "EF (%)", "EF < 50",
  "eGFR", "eGFR<50", "24hUP (g)", "24hUP≥5g", "Albumin (g/L)", "Alb<30",
  "treat", "Dara firstline", "Bort firstline", "Ven firstline"
)

baseline_vars <- intersect(desired_order, names(uni_df))
baseline_vars

tbl1 <- build_summary_table(
  df = uni_df,
  vars = baseline_vars,
  by = "productive_IgH",
  force_continuous = c("Involved Organs (n)")
)

tbl1

# =========================================================
# Baseline table by cardiac involvement
# =========================================================
desired_order <- c(
  "Sex", "Age", "Age≥65", "AL isotype (1=λ)", "productive_IgH", "t(11;14)",
  "Cardiac involvement", "Renal involvement", "Hepatic involvement",
  "BMPC (%)", "BMPC>10", "dFLC", "dFLC≥180",
  "eGFR", "eGFR<50", "24hUP (g)", "24hUP≥5g", "Albumin (g/L)", "Alb<30",
  "Dara firstline", "Bort firstline", "Ven firstline"
)

baseline_vars <- intersect(desired_order, names(uni_df))

tbl2 <- build_summary_table(
  df = uni_df,
  vars = baseline_vars,
  by = "Cardiac involvement"
)

tbl2

# =========================================================
# Create response variables
# =========================================================
uni_df <- uni_df %>%
  mutate(
    across(
      c(`3m HR`,`6m HR`,`12m HR`,`24m HR`),
      ~ ifelse(. %in% c("CR","VGPR","PR"), 1,
               ifelse(. %in% c("SD"), 0, NA)),
      .names = "{gsub(' HR','',.col)}_or"
    ),
    across(
      c(`3m HR`,`6m HR`,`12m HR`,`24m HR`),
      ~ ifelse(. %in% c("CR","VGPR"), 1,
               ifelse(. %in% c("PR","SD"), 0, NA)),
      .names = "{gsub(' HR','',.col)}_vgpr"
    ),
    across(
      c(`3m HR`,`6m HR`,`12m HR`,`24m HR`),
      ~ ifelse(. %in% c("CR"), 1,
               ifelse(. %in% c("VGPR","PR","SD"), 0, NA)),
      .names = "{gsub(' HR','',.col)}_cr_only"
    ),
    across(
      starts_with("cardiac_"),
      ~ ifelse(. %in% c("CR","VGPR","PR"), 1,
               ifelse(. %in% c("SD"), 0, NA)),
      .names = "car_{.col}_or"
    ),
    across(
      starts_with("cardiac_"),
      ~ ifelse(. %in% c("CR","VGPR"), 1,
               ifelse(. %in% c("PR","SD"), 0, NA)),
      .names = "car_{.col}_vgpr"
    )
  )

# =========================================================
# Response summary table by productive_IgH
# =========================================================
response_vars <- c(
  "best_hema_resp", "hema_or", "hema_vgpr", "hema_cr_only",
  "3m_or","6m_or","12m_or","24m_or",
  "3m_vgpr","6m_vgpr","12m_vgpr","24m_vgpr",
  "3m_cr_only","6m_cr_only","12m_cr_only","24m_cr_only",
  "car_cardiac_12m_or","car_cardiac_24m_or",
  "car_cardiac_12m_vgpr","car_cardiac_24m_vgpr"
)

response_vars <- intersect(response_vars, names(uni_df))

tbl3 <- build_summary_table(
  df = uni_df,
  vars = response_vars,
  by = "productive_IgH"
)

tbl3 <- tbl3 %>%
  mutate(
    variable = recode(
      variable,
      "best_hema_resp" = "Best Heme Response",
      "hema_cr_only" = "CR Rate",
      "hema_or" = "ORR",
      "hema_vgpr" = "≥VGPR rate"
    )
  )

# rename group columns if there are exactly 2 groups
stat_cols_tbl3 <- grep("^stat_", names(tbl3), value = TRUE)
if (length(stat_cols_tbl3) == 2) {
  tbl3 <- rename_stat_columns(
    tbl3,
    labels = c("LC only (N=407)", "IgH prod (N=373)")
  )
}

tbl3

# =========================================================
# Hematologic response stacked bar plot
# =========================================================
plot_long <- uni_df %>%
  filter(!is.na(`productive_IgH`)) %>%   
  dplyr::select(`productive_IgH`, `3m HR`, `6m HR`, `12m HR`, `24m HR`) %>%
  tidyr::pivot_longer(
    cols = c(`3m HR`, `6m HR`, `12m HR`, `24m HR`),
    names_to = "Time",
    values_to = "Response",
    names_pattern = "(.*) HR"
  ) %>%
  dplyr::mutate(
    Time = factor(Time, levels = c("3m", "6m", "12m", "24m")),
    Response = factor(Response, levels = c("PD","SD","PR","VGPR","CR"))
  )

plot_summary <- plot_long %>%
  filter(!is.na(Response)) %>%
  group_by(`productive_IgH`, Time, Response) %>%
  summarise(n = n(), .groups = "drop") %>%
  group_by(`productive_IgH`, Time) %>%
  mutate(
    total = sum(n),
    percent = n / total * 100
  ) %>%
  ungroup()

plot_n_orr <- plot_long %>%
  filter(!is.na(Response)) %>%
  group_by(`productive_IgH`, Time) %>%
  summarise(
    n = n(),
    ORR = mean(Response %in% c("CR","VGPR","PR")) * 100,
    .groups = "drop"
  )

ggplot(plot_summary, aes(x = Time, y = percent, fill = Response)) +
  geom_bar(stat = "identity", color = "black") +
  geom_text(
    aes(label = sprintf("%.0f%%", percent)),
    position = position_stack(vjust = 0.5),
    color = "white",
    size = 3.5,
    fontface = "bold"
  ) +
  geom_text(
    data = plot_n_orr,
    aes(x = Time, y = 104, label = sprintf("n = %d\nORR = %.1f%%", n, ORR)),
    inherit.aes = FALSE,
    size = 4,
    fontface = "bold",
    lineheight = 0.9
  ) +
  scale_fill_manual(
    values = c(
      "CR"   = "#1a9850",
      "VGPR" = "#91cf60",
      "PR"   = "#fee08b",
      "SD"   = "#fc8d59",
      "PD"   = "#d73027"
    ),
    name = "Hema Response"
  ) +
  facet_wrap(
    ~`productive_IgH`,
    labeller = labeller(
      `productive_IgH` = c(
        "0" = "LC",
        "1" = "IgH"
      )
    )
  ) +
  coord_cartesian(ylim = c(0, 106)) +
  theme_classic()

# =========================================================
# Survival
# =========================================================
uni_df <- uni_df %>%
  rename(productive_IgH = `productive_IgH`)

# by productive_IgH
fit_ttnt_igh <- survfit(
  Surv(TTNT_time, TTNT_event) ~ productive_IgH,
  data = uni_df
)

fit_os_igh <- survfit(
  Surv(os_time, os_event) ~ productive_IgH,
  data = uni_df
)

# by Mayo12
fit_ttnt_mayo <- survfit(
  Surv(TTNT_time, TTNT_event) ~ Mayo12,
  data = uni_df
)

fit_os_mayo <- survfit(
  Surv(os_time, os_event) ~ Mayo12,
  data = uni_df
)

p1 <- ggsurvplot(
  fit_ttnt_mayo, data = uni_df,
  title = "TTNT", xlab = "Time (months)", ylab = "Probability",
  conf.int = FALSE, pval = TRUE,
  risk.table = TRUE,
  risk.table.height = 0.25,
  risk.table.y.text = FALSE,
  risk.table.fontsize = 4,
  legend = "right",
  break.time.by = 24,
  ggtheme = theme_classic(base_size = 14)
)

p2 <- ggsurvplot(
  fit_os_mayo, data = uni_df,
  title = "OS", xlab = "Time (months)", ylab = "Probability",
  conf.int = FALSE, pval = TRUE,
  risk.table = TRUE,
  risk.table.height = 0.25,
  risk.table.y.text = FALSE,
  risk.table.fontsize = 4,
  legend = "right",
  break.time.by = 24,
  ggtheme = theme_classic(base_size = 14)
)

p1_combined <- ggarrange(p1$plot, p1$table, ncol = 1, heights = c(3, 1))
p2_combined <- ggarrange(p2$plot, p2$table, ncol = 1, heights = c(3, 1))
ggarrange(p1_combined, p2_combined, ncol = 2, nrow = 1)

# =========================================================
# Regression and association analysis
# =========================================================
uni_df$log_dFLC <- log(uni_df$dFLC)

df2 <- uni_df %>% 
  dplyr::select(`Cardiac involvement`, productive_IgH, log_dFLC) %>% 
  na.omit()

lm(log(`NT-ProBNP(pg/mL)`) ~ productive_IgH + log_dFLC, data = uni_df)

summary(glm(`Cardiac involvement` ~ productive_IgH, data = df2, family = binomial))
summary(glm(`Cardiac involvement` ~ log_dFLC, data = df2, family = binomial))
summary(glm(`Cardiac involvement` ~ productive_IgH + log_dFLC, data = df2, family = binomial))
summary(lm(log_dFLC ~ productive_IgH, data = df2))

m1 <- glm(
  `Cardiac involvement` ~ 
    Age +
    productive_IgH +
    `BMPC>10` +
    log_dFLC +
    `eGFR<50` +
    `Alb<30` +
    `Dara firstline`,
  data = uni_df,
  family = binomial
)

table_m1 <- build_regression_table(m1, exponentiate = TRUE)
table_m1

# =========================================================
# Subset check
# =========================================================
df3 <- subset(uni_df, productive_IgH == 1)
nrow(df3)

summary(glm(`Cardiac involvement` ~ log_dFLC, data = df2, family = binomial))

# =========================================================
# Mediation analysis
# =========================================================
med.model <- lm(log_dFLC ~ productive_IgH, data = df2)

out.model <- glm(
  `Cardiac involvement` ~ productive_IgH + log_dFLC,
  data = df2,
  family = binomial
)

med.out <- mediate(
  med.model,
  out.model,
  treat = "productive_IgH",
  mediator = "log_dFLC",
  boot = TRUE,
  sims = 1000
)

summary(med.out)
```

```r
library(dplyr)
library(tidyr)
library(purrr)
library(broom)
library(stringr)

# -----------------------------
# helper: pretty p-value
# -----------------------------
fmt_p <- function(p) {
  if (is.na(p)) return(NA_character_)
  if (p < 0.001) return("<0.001")
  sprintf("%.3f", p)
}

# -----------------------------
# helper: safe tests
# -----------------------------
get_p_cont <- function(data, var, by = "productive_IgH") {
  df <- data %>%
    dplyr::select(all_of(c(by, var))) %>%
    filter(!is.na(.data[[by]]), !is.na(.data[[var]]))

  if (nrow(df) == 0) return(NA_real_)
  if (length(unique(df[[by]])) < 2) return(NA_real_)
  if (length(unique(df[[var]])) < 2) return(NA_real_)

  out <- tryCatch({
    wilcox.test(df[[var]] ~ df[[by]])$p.value
  }, error = function(e) NA_real_)

  out
}

get_p_cat <- function(data, var, by = "productive_IgH") {
  df <- data %>%
    dplyr::select(all_of(c(by, var))) %>%
    filter(!is.na(.data[[by]]), !is.na(.data[[var]]))

  if (nrow(df) == 0) return(NA_real_)
  if (length(unique(df[[by]])) < 2) return(NA_real_)
  if (length(unique(df[[var]])) < 2) return(NA_real_)

  tab <- table(df[[var]], df[[by]])

  if (any(dim(tab) < 2)) return(NA_real_)

  out <- tryCatch({
    if (any(tab < 5)) {
      fisher.test(tab)$p.value
    } else {
      suppressWarnings(chisq.test(tab)$p.value)
    }
  }, error = function(e) NA_real_)

  out
}

# -----------------------------
# helper: detect variable type
# treat numeric with only 0/1 as categorical
# -----------------------------
detect_var_type <- function(x) {
  if (is.numeric(x)) {
    ux <- unique(x[!is.na(x)])
    if (length(ux) <= 2 && all(sort(ux) %in% c(0, 1))) {
      return("categorical")
    } else {
      return("continuous")
    }
  } else {
    return("categorical")
  }
}

# -----------------------------
# helper: summarize one continuous variable
# -----------------------------
summarise_continuous <- function(data, var, by = "productive_IgH", labels = NULL) {
  group_vals <- sort(unique(data[[by]][!is.na(data[[by]])]))

  stat_list <- lapply(group_vals, function(g) {
    x <- data %>%
      filter(.data[[by]] == g) %>%
      pull(all_of(var))

    tibble(
      group = as.character(g),
      value = if (all(is.na(x))) {
        NA_character_
      } else {
        sprintf(
          "%.1f (%.1f–%.1f)",
          median(x, na.rm = TRUE),
          min(x, na.rm = TRUE),
          max(x, na.rm = TRUE)
        )
      }
    )
  })

  stat_df <- bind_rows(stat_list) %>%
    pivot_wider(names_from = group, values_from = value)

  pval <- get_p_cont(data, var, by)

  tibble(
    Variable = if (!is.null(labels) && var %in% names(labels)) labels[[var]] else var,
    Level = ""
  ) %>%
    bind_cols(stat_df) %>%
    mutate(
      p_value = fmt_p(pval),
      sig = ifelse(!is.na(pval) & pval < 0.05, "*", "")
    )
}

# -----------------------------
# helper: summarize one categorical variable
# -----------------------------
summarise_categorical <- function(data, var, by = "productive_IgH", labels = NULL) {
  df <- data %>%
    filter(!is.na(.data[[var]]), !is.na(.data[[by]])) %>%
    mutate(
      .var = as.character(.data[[var]]),
      .by  = as.character(.data[[by]])
    )

  if (nrow(df) == 0) {
    return(
      tibble(
        Variable = if (!is.null(labels) && var %in% names(labels)) labels[[var]] else var,
        Level = NA_character_,
        p_value = NA_character_,
        sig = ""
      )
    )
  }

  group_vals <- sort(unique(df$.by))
  lvl_vals <- unique(df$.var)

  counts <- df %>%
    count(.var, .by, name = "n") %>%
    group_by(.by) %>%
    mutate(pct = 100 * n / sum(n)) %>%
    ungroup() %>%
    mutate(stat = sprintf("%d (%.1f%%)", n, pct)) %>%
    select(.var, .by, stat) %>%
    pivot_wider(names_from = .by, values_from = stat)

  pval <- get_p_cat(data, var, by)

  out <- counts %>%
    rename(Level = .var) %>%
    mutate(
      Variable = if (!is.null(labels) && var %in% names(labels)) labels[[var]] else var,
      p_value = ""
    ) %>%
    relocate(Variable, Level)

  if (nrow(out) > 0) {
    out$p_value[1] <- fmt_p(pval)
  }

  out <- out %>%
    mutate(
      sig = ""
    )

  if (nrow(out) > 0 && !is.na(pval) && pval < 0.05) {
    out$sig[1] <- "*"
  }

  out
}

# -----------------------------
# main table builder
# -----------------------------
make_summary_table <- function(data, vars, by = "productive_IgH", labels = NULL) {
  vars <- intersect(vars, names(data))

  pieces <- lapply(vars, function(v) {
    vtype <- detect_var_type(data[[v]])
    if (vtype == "continuous") {
      summarise_continuous(data, v, by, labels)
    } else {
      summarise_categorical(data, v, by, labels)
    }
  })

  out <- bind_rows(pieces)

  # add group Ns
  n_df <- data %>%
    filter(!is.na(.data[[by]])) %>%
    count(.data[[by]], name = "N") %>%
    mutate(group = as.character(.data[[by]]))

  group_vals <- sort(unique(data[[by]][!is.na(data[[by]])]))

  # rename group columns more nicely
  old_names <- intersect(as.character(group_vals), names(out))
  new_names <- paste0(old_names, " (N=", n_df$N[match(old_names, n_df$group)], ")")
  names(out)[match(old_names, names(out))] <- new_names

  out
}
```

```r
# ============================================================
# Mediation analysis + Nature-style visualization
# Works directly from med.out produced by mediation::mediate()
# ============================================================

library(dplyr)
library(ggplot2)
library(patchwork)
library(grid)
library(mediation)

# ------------------------------------------------------------
# 1. Run mediation analysis
# ------------------------------------------------------------
uni_df$log_dFLC <- log(uni_df$dFLC)

df2 <- uni_df %>%
  dplyr::select(`Cardiac involvement`, productive_IgH, log_dFLC) %>%
  na.omit()

# make sure treatment is coded consistently
# if productive_IgH is character/factor, this is usually safer:
# df2$productive_IgH <- factor(df2$productive_IgH, levels = c("0", "1"))

med.model <- lm(
  log_dFLC ~ productive_IgH,
  data = df2
)

out.model <- glm(
  `Cardiac involvement` ~ productive_IgH + log_dFLC,
  data = df2,
  family = binomial
)

med.out <- mediate(
  model.m  = med.model,
  model.y  = out.model,
  treat    = "productive_IgH",
  mediator = "log_dFLC",
  boot     = TRUE,
  sims     = 1000
)

summary(med.out)

# ------------------------------------------------------------
# 2. Extract mediation results correctly from med.out
# ------------------------------------------------------------
# For a standard binary treatment mediation analysis, med.out stores:
# ACME = d.avg
# ADE  = z.avg
# Total effect = tau.coef
# Proportion mediated = n.avg
#
# Bootstrap CIs:
# ACME: d.avg.ci
# ADE:  z.avg.ci
# Total: tau.ci
# Prop mediated: n.avg.ci

med_df <- data.frame(
  effect = c("ACME", "ADE", "Total effect", "Prop. mediated"),
  estimate = c(
    unname(med.out$d.avg),
    unname(med.out$z.avg),
    unname(med.out$tau.coef),
    unname(med.out$n.avg)
  ),
  lower = c(
    unname(med.out$d.avg.ci[1]),
    unname(med.out$z.avg.ci[1]),
    unname(med.out$tau.ci[1]),
    unname(med.out$n.avg.ci[1])
  ),
  upper = c(
    unname(med.out$d.avg.ci[2]),
    unname(med.out$z.avg.ci[2]),
    unname(med.out$tau.ci[2]),
    unname(med.out$n.avg.ci[2])
  )
)

eff_df <- med_df %>%
  filter(effect %in% c("ACME", "ADE", "Total effect")) %>%
  mutate(effect = factor(effect, levels = c("Total effect", "ADE", "ACME")))

prop_df <- med_df %>%
  filter(effect == "Prop. mediated")

acme_val  <- med_df$estimate[med_df$effect == "ACME"]
ade_val   <- med_df$estimate[med_df$effect == "ADE"]
total_val <- med_df$estimate[med_df$effect == "Total effect"]
prop_val  <- med_df$estimate[med_df$effect == "Prop. mediated"]

acme_ci  <- med_df %>% filter(effect == "ACME") %>% dplyr::select(lower, upper)
ade_ci   <- med_df %>% filter(effect == "ADE") %>% dplyr::select(lower, upper)
total_ci <- med_df %>% filter(effect == "Total effect") %>% dplyr::select(lower, upper)
prop_ci  <- med_df %>% filter(effect == "Prop. mediated") %>% dplyr::select(lower, upper)

# Optional p-values, if present in med.out
get_safe <- function(x) {
  if (is.null(x)) return(NA_real_)
  unname(x)
}

p_acme <- get_safe(med.out$d.avg.p)
p_ade  <- get_safe(med.out$z.avg.p)
p_tot  <- get_safe(med.out$tau.p)
p_prop <- get_safe(med.out$n.avg.p)

fmt_num <- function(x, digits = 3) sprintf(paste0("%.", digits, "f"), x)
fmt_pct <- function(x, digits = 1) sprintf(paste0("%.", digits, "f%%"), x * 100)
fmt_p <- function(p) {
  if (is.na(p)) return("p = NA")
  if (p < 0.001) return("p < 0.001")
  paste0("p = ", sprintf("%.3f", p))
}

# ------------------------------------------------------------
# 3. Panel A: mediation diagram
# ------------------------------------------------------------
pA <- ggplot() +
  xlim(0, 10) +
  ylim(0, 6.6) +
  theme_void() +

  # nodes
  annotate(
    "label", x = 1.5, y = 3.0, label = "X\nproductive_IgH",
    size = 5.0, label.size = 0.4, fill = "white"
  ) +
  annotate(
    "label", x = 5.0, y = 4.9, label = "M\nlog(dFLC)",
    size = 5.0, label.size = 0.4, fill = "white"
  ) +
  annotate(
    "label", x = 8.5, y = 3.0, label = "Y\nCardiac involvement",
    size = 5.0, label.size = 0.4, fill = "white"
  ) +

  # arrows
  annotate(
    "segment", x = 2.3, xend = 4.1, y = 3.5, yend = 4.5,
    linewidth = 0.8,
    arrow = arrow(length = unit(0.18, "inches"), type = "closed")
  ) +
  annotate(
    "segment", x = 5.9, xend = 7.7, y = 4.5, yend = 3.5,
    linewidth = 0.8,
    arrow = arrow(length = unit(0.18, "inches"), type = "closed")
  ) +
  annotate(
    "segment", x = 2.3, xend = 7.7, y = 2.35, yend = 2.35,
    linewidth = 0.8,
    arrow = arrow(length = unit(0.18, "inches"), type = "closed")
  ) +

  # path labels
  annotate("text", x = 3.2, y = 4.62, label = "a path", size = 4.4) +
  annotate("text", x = 6.8, y = 4.62, label = "b path", size = 4.4) +
  annotate("text", x = 5.0, y = 2.72, label = "Direct effect (ADE)", size = 4.2) +

  # proportion mediated
  annotate(
    "text", x = 5.0, y = 5.95,
    label = paste0(fmt_pct(prop_val, 1), " mediated"),
    size = 4.3, fontface = "bold"
  ) +

  # result summary
  annotate(
    "text", x = 5.0, y = 1.02,
    label = paste0(
      "Indirect effect (ACME) = ", fmt_num(acme_val), 
      " [", fmt_num(acme_ci$lower), ", ", fmt_num(acme_ci$upper), "]\n",
      "Direct effect (ADE) = ", fmt_num(ade_val), 
      " [", fmt_num(ade_ci$lower), ", ", fmt_num(ade_ci$upper), "]\n",
      "Total effect = ", fmt_num(total_val), 
      " [", fmt_num(total_ci$lower), ", ", fmt_num(total_ci$upper), "]\n",
      "Proportion mediated = ", fmt_pct(prop_val, 1),
      " [", fmt_pct(prop_ci$lower, 1), ", ", fmt_pct(prop_ci$upper, 1), "]"
    ),
    size = 4.0, lineheight = 1.15
  ) +

  annotate(
    "text", x = 0.2, y = 6.45, label = "A",
    hjust = 0, vjust = 1, size = 7, fontface = "bold"
  )

# ------------------------------------------------------------
# 4. Panel B: forest plot for ACME/ADE/Total
# ------------------------------------------------------------
label_df <- eff_df %>%
  mutate(
    label_txt = sprintf("%.3f", estimate),
    label_x = min(lower, na.rm = TRUE) - 0.06 * diff(range(c(lower, upper), na.rm = TRUE))
  )

pB <- ggplot(eff_df, aes(x = estimate, y = effect)) +
  geom_vline(xintercept = 0, linetype = "dashed", linewidth = 0.5, color = "grey50") +
  geom_errorbarh(aes(xmin = lower, xmax = upper), height = 0.16, linewidth = 0.8) +
  geom_point(size = 3.2, shape = 21, fill = "black", color = "black", stroke = 0.6) +
  geom_text(
    data = label_df,
    aes(x = label_x, y = effect, label = label_txt),
    inherit.aes = FALSE,
    hjust = 1,
    size = 3.8
  ) +
  scale_x_continuous(expand = expansion(mult = c(0.20, 0.08))) +
  labs(x = "Effect size (95% bootstrap CI)", y = NULL) +
  theme_classic(base_size = 13) +
  theme(
    axis.text.y = element_text(size = 12, color = "black"),
    axis.text.x = element_text(size = 11, color = "black"),
    axis.title.x = element_text(size = 12),
    plot.margin = margin(10, 15, 10, 10)
  ) +
  annotate(
    "text", x = -Inf, y = Inf, label = "B",
    hjust = -0.7, vjust = 1.7, size = 7, fontface = "bold"
  )

# ------------------------------------------------------------
# 5. Panel C: proportion mediated with correct CI scale
# ------------------------------------------------------------
ymin_c <- min(0, prop_df$lower, prop_df$estimate, na.rm = TRUE)
ymax_c <- max(0, prop_df$upper, prop_df$estimate, na.rm = TRUE)
ypad_c <- 0.08 * (ymax_c - ymin_c)
if (ypad_c == 0) ypad_c <- 0.1

pC <- ggplot(prop_df, aes(x = effect, y = estimate)) +
  geom_col(width = 0.52, fill = "grey20") +
  geom_errorbar(aes(ymin = lower, ymax = upper), width = 0.12, linewidth = 0.8) +
  geom_hline(yintercept = 0, linewidth = 0.5) +
  coord_cartesian(ylim = c(ymin_c - ypad_c, ymax_c + ypad_c)) +
  labs(x = NULL, y = "Proportion mediated") +
  theme_classic(base_size = 13) +
  theme(
    axis.text.x = element_text(size = 11, color = "black"),
    axis.text.y = element_text(size = 11, color = "black"),
    axis.title.y = element_text(size = 12),
    plot.margin = margin(10, 10, 10, 10)
  ) +
  annotate(
    "text", x = 1, y = prop_df$estimate + 0.04 * (ymax_c - ymin_c + 1e-8),
    label = fmt_pct(prop_df$estimate, 1),
    size = 4
  ) +
  annotate(
    "text", x = -Inf, y = Inf, label = "C",
    hjust = -0.7, vjust = 1.7, size = 7, fontface = "bold"
  )

# ------------------------------------------------------------
# 6. Combine panels
# ------------------------------------------------------------
final_plot <- pA + pB
  plot_annotation(
    title = "Causal mediation analysis",
    theme = theme(
      plot.title = element_text(size = 15, face = "bold", hjust = 0.5)
    )
  )

final_plot

# ------------------------------------------------------------
# 7. Save
# ------------------------------------------------------------
ggsave(
  "mediation_nature_style.pdf",
  final_plot,
  width = 10,
  height = 5,
  useDingbats = FALSE
)

ggsave(
  "mediation_nature_style.png",
  final_plot,
  width = 10,
  height = 5,
  dpi = 400
)
```
